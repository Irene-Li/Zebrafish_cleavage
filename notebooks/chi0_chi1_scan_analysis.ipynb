```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
import os, sys

sys.path.append(str(Path(os.getcwd()).parent / "src"))

import importlib
import utils
importlib.reload(utils)
```

```python
# Grid parameters (must match chi0_chi1_scan.py)
n_chi = 50
chi0_values = np.exp(np.linspace(np.log(0.01), np.log(1), n_chi))
chi1_values = np.exp(np.linspace(np.log(0.01), np.log(1), n_chi))
# chi0_values = chi0_values[:-2]

data_dir = Path("../sim_data/acto_myo")

# Load all available data files
data = {}
for chi0 in chi0_values:
    for chi1 in chi1_values:
        filename = data_dir / f"chi0_{chi0:.4f}_chi1_{chi1:.4f}.npz"
        if filename.exists():
            data[(chi0, chi1)] = np.load(filename)

print(f"Loaded {len(data)} simulations")


extract_states(data[(chi0_values[0], chi1_values[0])])
```

```python
# Extract states
states = []
for chi0 in chi0_values:
    for chi1 in chi1_values:
        if (chi0, chi1) in data:
            states.append(extract_states(data[(chi0, chi1)], shear_range=6))
        else:
            states.append([np.nan, np.nan, np.nan, np.nan, np.nan])

states = np.array(states).reshape(len(chi0_values), len(chi1_values), 5)
```

```python
from matplotlib.colors import ListedColormap

# Figure 1: Density ratio, shear pattern, and Q
fig1, axes1 = plt.subplots(1, 3, figsize=(18, 5))
names1 = [r'$\rho_\mathrm{min}/\rho_\mathrm{edge}$', r'shear pattern', r'$Q(0, 0)$']
ranges1 = [(0, 1), (-1, 1), (-1, 1)]

shear_colors = ['#276419', 'white', '#8e0152']
shear_cmap = ListedColormap(shear_colors)
cmaps1 = ['Blues', shear_cmap, 'PiYG']

for i, ax in enumerate(axes1):
    c = ax.pcolor(chi1_values, chi0_values, states[:, :, i],
                  cmap=cmaps1[i], vmin=ranges1[i][0], vmax=ranges1[i][1])
    if i == 0:
        ax.contour(chi1_values, chi0_values, states[:, :, i], levels=[0.4], colors='k', linestyles='dashed')
    ax.set_xlabel(r'$\chi_1$', fontsize=16)
    ax.set_ylabel(r'$\chi_0$', fontsize=16)
    ax.set_xscale('log')
    ax.set_yscale('log')
    ax.set_title(names1[i], fontsize=16)
    ax.tick_params(axis='both', which='major', labelsize=14)
    cbar = fig1.colorbar(c, ax=ax)
    cbar.ax.tick_params(labelsize=14)
    if i == 0:
        cbar.set_ticks([0, 0.5, 1.0])
        cbar.set_ticklabels(['0', '0.5', '1.0'])
    elif i == 1:
        cbar.set_ticks([-1, 0, 1])
    elif i == 2:
        cbar.set_ticks([-1, -0.5, 0, 0.5, 1])
        cbar.set_ticklabels(['< -1', '-0.5', '0', '0.5', '> 1'])

fig1.tight_layout()
fig1.savefig('../figures/chi0_chi1_summary_2d_part1.png', dpi=500, bbox_inches='tight')
plt.show()
```

```python

# Figure 2: Min and Max compression rates
fig2, axes2 = plt.subplots(1, 2, figsize=(12, 5))
names2 = [r'$\min(dvx/dx)$', r'$\max(dvx/dx)$']
ranges2 = [(-1, 1), (-1, 1)]
cmaps2 = ['coolwarm', 'coolwarm']
state_indices = [3, 4]

for i, ax in enumerate(axes2):
    idx = state_indices[i]
    c = ax.pcolor(chi1_values, chi0_values, states[:, :, idx],
                  cmap=cmaps2[i], vmin=ranges2[i][0], vmax=ranges2[i][1])
    if i == 0:
        ax.contour(chi1_values, chi0_values, states[:, :, idx], levels=[-0.3], colors='k', linestyles='dashed')
    elif i == 1:
        ax.contour(chi1_values, chi0_values, states[:, :, idx], levels=[0.2], colors='k', linestyles='dashed')
    ax.set_xlabel(r'$\chi_1$', fontsize=16)
    ax.set_ylabel(r'$\chi_0$', fontsize=16)
    ax.set_xscale('log')
    ax.set_yscale('log')
    ax.set_title(names2[i], fontsize=16)
    ax.tick_params(axis='both', which='major', labelsize=14)
    cbar = fig2.colorbar(c, ax=ax)
    cbar.ax.tick_params(labelsize=14)

fig2.tight_layout()
fig2.savefig('../figures/chi0_chi1_summary_2d_part2.png', dpi=500, bbox_inches='tight')
plt.show()
```
